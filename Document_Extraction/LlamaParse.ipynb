```python
from llama_cloud_services import LlamaParse

parser = LlamaParse(
    api_key=dbutils.secrets.get(scope = "gen_ai", key = "LLAMA_CLOUD_API_KEY"),  # can also be set in your env as LLAMA_CLOUD_API_KEY
    num_workers=4,       # if multiple files passed, split in `num_workers` API calls
    verbose=True,
    language="en",       # optionally define a language, default=en
    auto_mode=True,
    extract_layout=True, #it will include bounding boxes
)
```

```python
volume_path = "/Volumes/data/extraction/unstructed/raw/"
```

```python
# sync
#docs=[volume_path+"1098T.pdf",volume_path+"W-2_Form_2020.pdf"]
docs = volume_path+"1098T.pdf"
result = parser.parse(docs)
```

```python
# get the llama-index markdown documents
markdown_documents = result.get_markdown_documents(split_by_page=True)
markdown_documents
```

```python
# get the llama-index text documents
text_documents = result.get_text_documents(split_by_page=False)
text_documents
```

```python
# access the raw job result
# Items will vary based on the parser configuration
for page in result.pages:
    print(page.md)
```
